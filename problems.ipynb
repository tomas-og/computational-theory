##### Problem 1: Representing SHA-256 Data

```python
MASK = 0xFFFFFFFF  #keeps a Python int within 32 bits


#32-bit words are plain ints, masked to 32 bits.
#Python ints never overflow, so we mask after +, <<, and ~.
def add32(*args):
    return sum(args) & MASK  # addition mod 2**32


def rotr(x, n):
    return ((x >> n) | (x << (32 - n))) & MASK  # rotate right by n bits

#Padding function for SHA-256.
#The message is padded with a 1 bit then 0 bits until
#the lenght of the message is 448 mod 512
#Then the lenght of the original message is appended as a 64 bit
#big endian interger. The result is a multiple of 512 bits.)
def pad (msg: bytes) -> bytes:
    bitlenght = len(msg) * 8
    msg += b"\x80"
    msg += b"\x00" * ((56 - len(msg)) % 64)
    msg += bitlenght.to_bytes(8, "big")
    return msg

#Split the padded message into 64 byte blocks
def convertblocks(padded: bytes):
    return [padded[i:i + 64] for i in range(0, len(padded), 64)]

#Convert a 64 byte block into 16 32 bit words
#SHA 256 is big edian so the first byte is the most immportant
def blocktowords(block: bytes):
    return [int.from_bytes(block[i:i + 4], "big") for i in range(0, 64, 4)]

#Debugging Check
assert add32(0xFFFFFFFF, 1) == 0
assert rotr(1, 1) == 0x80000000
print("The code is running: 32 bit word helpers is working")

padded = pad(b"abc")
assert len(padded) % 64 == 0
blocks = convertblocks(padded)
assert len(blocks) == 1 and len(blocks[0]) == 64
print("The code is running: padding and block splitting is working")

words = blocktowords(blocks[0])
assert len(words) == 16
assert words[0] == 0x61626380 #abc followed by 1 bit and 0 bits
assert words[15] == 0x18 #messange lenght is 24 bits = 0x18
assert (0x61626380).to_bytes(4, "big") == b"abc\x80"
print("The code is running: block to words is working")
```
